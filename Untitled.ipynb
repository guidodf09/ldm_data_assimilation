```python
import torch
import torch.nn as nn
import os


def train_model(model, criterion, optimizer, scheduler, train_loader, val_loader,
                device, model_save_path, num_epochs=100, patience=10):
    best_val_loss = float('inf')
    epochs_no_improve = 0
    best_model_wts = None
    train_losses = []
    val_losses = []

    for epoch in range(num_epochs):
        # -----------------------------
        # Training
        # -----------------------------
        model.train()
        train_loss = 0.0
        for geo_batch, rate_batch in train_loader:
            geo_batch  = geo_batch.to(device)
            rate_batch = rate_batch.to(device)
            optimizer.zero_grad()
            outputs = model(geo_batch)
            loss = criterion(outputs, rate_batch)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * geo_batch.size(0)
        train_loss /= len(train_loader.dataset)
        train_losses.append(train_loss)

        # -----------------------------
        # Validation
        # -----------------------------
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for geo_batch, rate_batch in val_loader:
                geo_batch  = geo_batch.to(device)
                rate_batch = rate_batch.to(device)
                outputs = model(geo_batch)
                loss = criterion(outputs, rate_batch)
                val_loss += loss.item() * geo_batch.size(0)
        val_loss /= len(val_loader.dataset)
        val_losses.append(val_loss)

        if scheduler is not None:
            scheduler.step(val_loss)
        print(f"Epoch {epoch+1}/{num_epochs} | Train Loss: {train_loss:.6f} | Val Loss: {val_loss:.6f}")

        # -----------------------------
        # Early Stopping
        # -----------------------------
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            epochs_no_improve = 0
            best_model_wts = model.state_dict()
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f"Early stopping triggered after {epoch+1} epochs")
                break

    if best_model_wts is not None:
        model.load_state_dict(best_model_wts)
    os.makedirs(os.path.dirname(model_save_path), exist_ok=True)
    torch.save(model.state_dict(), model_save_path)
    print(f"Trained model saved to {model_save_path}")

    return model, train_losses, val_losses
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset


# -----------------------------
# 3D Conv Block
# -----------------------------
class Conv3DBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size=3, stride=1, padding=1):
        super().__init__()
        self.block = nn.Sequential(
            nn.Conv3d(in_channels, out_channels, kernel_size, stride, padding),
            nn.BatchNorm3d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.block(x)


# -----------------------------
# Geo (3-channel) → Rates Model
#   channels: [poro, perm, ar]   (facies is NOT fed to the model)
# -----------------------------
class GeoToRatesModel(nn.Module):
    def __init__(
        self,
        input_shape=(3, 128, 128, 32),   # 3 property channels
        conv_channels=(32, 64, 128),
        latent_dim=256,
        time_steps=30,
        lstm_hidden_size=128,
        lstm_layers=2,
        output_dim=15,
        bidirectional=False,
        dropout_rate=0.3,
    ):
        super().__init__()

        self.time_steps = time_steps
        self.bidirectional = bidirectional

        # CNN Encoder
        encoder_layers = []
        in_ch = input_shape[0]
        for out_ch in conv_channels:
            encoder_layers.append(Conv3DBlock(in_ch, out_ch, stride=1))
            encoder_layers.append(nn.MaxPool3d(kernel_size=2))
            in_ch = out_ch

        self.encoder = nn.Sequential(*encoder_layers)
        self.spatial_pool = nn.AdaptiveAvgPool3d((1, 1, 1))
        self.flatten = nn.Flatten()

        self.linear_latent = nn.Sequential(
            nn.Linear(conv_channels[-1], latent_dim),
            nn.ReLU(),
            nn.Dropout(p=dropout_rate)
        )

        # Learned Time Embeddings
        self.time_embedding = nn.Embedding(time_steps, latent_dim)

        # LSTM Decoder (input = repeated latent + time embedding)
        self.decoder = nn.LSTM(
            input_size=latent_dim * 2,
            hidden_size=lstm_hidden_size,
            num_layers=lstm_layers,
            batch_first=True,
            bidirectional=bidirectional,
            dropout=dropout_rate if lstm_layers > 1 else 0.0
        )

        decoder_output_dim = lstm_hidden_size * (2 if bidirectional else 1)
        self.output_layer = nn.Linear(decoder_output_dim, output_dim)

        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, (nn.Conv3d, nn.Linear)):
                nn.init.kaiming_normal_(m.weight)
                if m.bias is not None:
                    nn.init.zeros_(m.bias)

    def forward(self, x):
        """
        x : (N, 3, H, W, D)  -- [poro, perm, ar] (facies already removed)
        returns: (N, T, output_dim)
        """
        N = x.size(0)

        x = self.encoder(x)
        x = self.spatial_pool(x)
        x = self.flatten(x)
        latent = self.linear_latent(x)                 # (N, latent_dim)

        t_idx = torch.arange(self.time_steps, device=latent.device)
        t_emb = self.time_embedding(t_idx)             # (T, latent_dim)
        t_emb = t_emb.unsqueeze(0).repeat(N, 1, 1)     # (N, T, latent_dim)

        latent_seq = latent.unsqueeze(1).repeat(1, self.time_steps, 1)
        decoder_input = torch.cat([latent_seq, t_emb], dim=-1)

        rnn_out, _ = self.decoder(decoder_input)
        return self.output_layer(rnn_out)              # (N, T, output_dim)


# -----------------------------
# Dataset (no parameter vector)
# -----------------------------
class GeoRatesDataset(Dataset):
    def __init__(self, geo_data, rates_data):
        self.geo_data   = geo_data      # (N, 3, H, W, D)
        self.rates_data = rates_data    # (N, T, F)

    def __len__(self):
        return len(self.geo_data)

    def __getitem__(self, idx):
        return self.geo_data[idx], self.rates_data[idx]
```

```python
import os
import numpy as np
import h5py
import torch
import torch.nn as nn
import pandas as pd
from torch.utils.data import DataLoader, random_split
# from models import GeoRatesDataset, GeoToRatesModel
# from train import train_model
import time
import optuna
import json
from hyperopt import create_objective
from sklearn.preprocessing import StandardScaler, MinMaxScaler

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')


def main(mode="train",
         dataset_dir=".",
         dataset_file="paired_dataset_combined_chunk000.h5",
         rates_file="all_data_all_times.npy",
         case_name="case0",
         pics_name="pics"):
    # -----------------------------
    # Directory Setup
    # -----------------------------
    case_dir = os.path.join(os.getcwd(), case_name)
    pics_dir = os.path.join(case_dir, pics_name)
    os.makedirs(case_dir, exist_ok=True)
    os.makedirs(pics_dir, exist_ok=True)

    # -----------------------------
    # Load Data  (X has 4 channels: [facies, poro, perm, ar])
    # -----------------------------
    with h5py.File(os.path.join(dataset_dir, dataset_file), 'r') as f:
        X_loaded = np.array(f["X"])                 # (N, 128, 128, 32, 4)

    rates_loaded = np.load(os.path.join(dataset_dir, rates_file))  # (N, T, F)

    # channels-last -> channels-first, AND drop facies (channel 0).
    # keep only [poro, perm, ar] -> (N, 3, 128, 128, 32)
    X_loaded = np.transpose(X_loaded, (0, 4, 1, 2, 3))[:, 1:]
    X_loaded[:, 1] = np.log(X_loaded[:, 1])   # log-transform perm channel

    # -----------------------------
    # Shuffle (same permutation across both)
    # -----------------------------
    np.random.seed(42)
    random_order = np.random.permutation(rates_loaded.shape[0])
    rates_loaded = rates_loaded[random_order]
    X_loaded     = X_loaded[random_order]

    # -----------------------------
    # Normalize
    # -----------------------------
    normalization_type = 'standard'
    Scaler = StandardScaler if normalization_type == 'standard' else MinMaxScaler

    N, T, F = rates_loaded.shape

    rate_scaler      = Scaler()
    rates_normalized = torch.tensor(
        rate_scaler.fit_transform(rates_loaded.reshape(-1, F)).reshape(N, T, F),
        dtype=torch.float32
    )

    # Per-channel normalization over batch + spatial dims (3 channels: poro, perm, ar)
    geo = torch.tensor(X_loaded, dtype=torch.float32)            # (N, 3, X, Y, Z)
    if normalization_type == 'standard':
        ch_mean = geo.mean(dim=(0, 2, 3, 4), keepdim=True)
        ch_std  = geo.std(dim=(0, 2, 3, 4), keepdim=True) + 1e-6
        geo_normalized = (geo - ch_mean) / ch_std
        geo_stats = {'mean': ch_mean, 'std': ch_std}
    else:
        ch_min = geo.amin(dim=(0, 2, 3, 4), keepdim=True)
        ch_max = geo.amax(dim=(0, 2, 3, 4), keepdim=True)
        geo_normalized = (geo - ch_min) / (ch_max - ch_min + 1e-6)
        geo_stats = {'min': ch_min, 'max': ch_max}
    torch.save(geo_stats, os.path.join(case_dir, 'geo_channel_stats.pt'))

    # -----------------------------
    # Dataset and DataLoaders
    # -----------------------------
    total_samples = len(geo_normalized)
    train_size    = int(0.8 * total_samples)
    val_size      = int(0.1 * total_samples)
    test_size     = total_samples - train_size - val_size

    generator = torch.Generator().manual_seed(42)

    train_dataset, val_dataset, test_dataset = random_split(
        GeoRatesDataset(geo_normalized, rates_normalized),
        [train_size, val_size, test_size],
        generator=generator
    )

    batch_size   = 32
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    val_loader   = DataLoader(val_dataset,   batch_size=batch_size, shuffle=False)
    test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False)

    # -----------------------------
    # Modes
    # -----------------------------
    if mode == 'hyperopt':
        objective = create_objective(train_loader, val_loader, time_steps=T, output_dim=F)
        study = optuna.create_study(
            direction="minimize",
            pruner=optuna.pruners.MedianPruner(n_warmup_steps=5)
        )
        study.optimize(objective, n_trials=100)
        print("Best trial:")
        print(study.best_trial)

    elif mode == 'train':
        best_params = {
            'conv_channels':    (16, 32, 64, 128),
            'latent_dim':       512,
            'lstm_hidden_size': 64,
            'lstm_layers':      1,
            'bidirectional':    False,
            'dropout':          0.1155930716554599,
            'lr':               0.002038540481683897,
            'weight_decay':     2.1669376391175675e-06,
            'use_huber':        True,
            'grad_clip':        1.0,
        }

        model = GeoToRatesModel(
            conv_channels=best_params["conv_channels"],
            latent_dim=best_params["latent_dim"],
            time_steps=T,
            lstm_hidden_size=best_params["lstm_hidden_size"],
            lstm_layers=best_params["lstm_layers"],
            bidirectional=best_params["bidirectional"],
            output_dim=F,
            dropout_rate=best_params['dropout'],
        ).to(device)

        criterion = nn.HuberLoss(delta=1.0) if best_params['use_huber'] else nn.MSELoss()
        optimizer = torch.optim.Adam(
            model.parameters(),
            lr=best_params['lr'],
            weight_decay=best_params['weight_decay']
        )
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer, mode='min', factor=0.5, patience=10, verbose=True
        )

        start_time = time.time()
        model, train_losses, val_losses = train_model(
            model, criterion, optimizer, scheduler,
            train_loader, val_loader, device,
            model_save_path=os.path.join(case_dir, "model_trained.pth"),
            num_epochs=1000, patience=30
        )
        total_time = time.time() - start_time

        print(f"Training complete.")
        print(f"Total training time: {total_time:.2f} seconds")
        print(f"Average time per epoch: {total_time / len(train_losses):.2f} seconds")

        np.save(os.path.join(case_dir, 'train_losses.npy'), np.array(train_losses))
        np.save(os.path.join(case_dir, 'val_losses.npy'),   np.array(val_losses))
        pd.DataFrame({"train_loss": train_losses, "val_loss": val_losses}).to_csv(
            os.path.join(case_dir, 'loss_history.csv'), index_label='epoch'
        )
        with open(os.path.join(case_dir, "model_config.json"), "w") as f:
            json.dump(best_params, f, indent=4)

    else:
        raise ValueError(f"Invalid mode '{mode}'. Use 'train' or 'hyperopt'.")
```
